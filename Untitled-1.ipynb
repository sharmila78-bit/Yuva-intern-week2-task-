```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# Set visualization style
sns.set_theme(style="whitegrid")

# 1. Load Dataset
df = sns.load_dataset('titanic')

# Print summary stats
print("--- DATASET SHAPE & SUMMARY ---")
print(df.info())
print("\n--- DESCRIPTIVE STATISTICS ---")
print(df.describe())

# Plot 1: Survival Count by Sex (Barplot)
plt.figure(figsize=(6, 4))
sns.countplot(x='sex', hue='survived', data=df, palette='Set2')
plt.title('Survival Count by Gender')
plt.xlabel('Gender')
plt.ylabel('Passenger Count')
plt.legend(['Did Not Survive', 'Survived'])
plt.tight_layout()
plt.savefig('plot1_survival_by_gender.png')
plt.show()

# Plot 2: Age Distribution by Survival (KDE / Histogram)
plt.figure(figsize=(7, 4))
sns.histplot(
    data=df,
    x='age',
    hue='survived',
    kde=True,
    element='step',
    palette='coolwarm',
)
plt.title('Age Distribution and Survival Pattern')
plt.xlabel('Age')
plt.ylabel('Density / Count')
plt.tight_layout()
plt.savefig('plot2_age_distribution.png')
plt.show()

# Plot 3: Passenger Class vs Fare (Boxplot for Outlier & Spread)
plt.figure(figsize=(7, 4))
sns.boxplot(x='pclass', y='fare', data=df, palette='Blues')
plt.title('Fare Distribution across Passenger Classes')
plt.xlabel('Passenger Class (1st, 2nd, 3rd)')
plt.ylabel('Ticket Fare')
plt.ylim(0, 300)  # Zoomed for readability
plt.tight_layout()
plt.savefig('plot3_pclass_fare_box.png')
plt.show()

# Plot 4: Correlation Heatmap for Numerical Features
plt.figure(figsize=(6, 5))
numeric_cols = df.select_dtypes(include=['number'])
sns.heatmap(numeric_cols.corr(), annot=True, cmap='mako', fmt='.2f')
plt.title('Feature Correlation Heatmap')
plt.tight_layout()
plt.savefig('plot4_correlation_heatmap.png')
plt.show()
```
